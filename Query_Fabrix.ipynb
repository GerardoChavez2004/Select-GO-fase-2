```python
%sql
--Fabricio
WITH tickets AS (
  SELECT id_venta, id_supermercado, id_canal_venta, SUM(subtotal_linea) AS total_ticket
  FROM selectgo_fase2.gold.fact_ventas
  GROUP BY id_venta, id_supermercado, id_canal_venta
)
SELECT t.nombre_supermercado, c.nombre_canal,
       ROUND(AVG(tk.total_ticket),2) AS ticket_promedio,
       RANK() OVER (ORDER BY AVG(tk.total_ticket) DESC) AS ranking
FROM tickets tk
JOIN selectgo_fase2.gold.dim_tienda t ON t.id_supermercado = tk.id_supermercado
JOIN selectgo_fase2.gold.dim_canal c ON c.id_canal_venta = tk.id_canal_venta
GROUP BY t.nombre_supermercado, c.nombre_canal
ORDER BY ticket_promedio DESC; 
```
